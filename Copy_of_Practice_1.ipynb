```python
income = 5000
debt = 1000
dtiratio = debt/income
print(f"debt to income ration is {dtiratio}")
```

```python
import matplotlib.pyplot as plt
lables =['Income', 'Debt']
values = [income, debt]

fig, ax = plt.subplots()
ax.bar(lables, values, color=['green', 'red'])
ax.set_title('Debt to Income RAtion')
ax.set_ylabel("Amount  ($)")
plt.show()
```
